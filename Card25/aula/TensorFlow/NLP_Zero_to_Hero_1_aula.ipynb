```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
sentences = { # criando as sentenças(A sequencia de frases que vão ser tokenizadas e transformadas)
    'I love my dog', # Eu amo meu cachorro
    'I love my cat', # Eu amo meu gato
    'You love my dog!', # Você adora o meu cachorro!
    'Do you think my dog is amazing?' # Você acha que meu cachorro é incrível?
}
```

```python
tokenizar = Tokenizer(num_words=100) # criando o tokenizador, num_words representa o numero máximo de palavras a serem mantidas
tokenizar.fit_on_texts(sentences)
word_index = tokenizar.word_index # criando o dicionário de palavras
print(word_index) # imprimindo o dicionário de palavras
```
