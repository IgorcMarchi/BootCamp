```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
sentences = { # criando as sentenças(A sequencia de frases que vão ser tokenizadas e transformadas)
    'I love my dog', # Eu amo meu cachorro
    'I love my cat', # Eu amo meu gato
    'You love my dog!', # Você adora o meu cachorro!
    'Do you think my dog is amazing?' # Você acha que meu cachorro é incrível?
}
```

```python
tokenizer = Tokenizer(num_words=100, oov_token="<OOV>") # criando o tokenizador, num_words representa o numero máximo de palavras a serem mantidas
tokenizer.fit_on_texts(sentences)
word_index = tokenizer.word_index # criando o dicionário de palavras
```

```python
sequences = tokenizer.texts_to_sequences(sentences) # criando a sequência de palavras
padded = pad_sequences(sequences, padding='post') # criando o padding das sequências, é colocado o padding='post' para deixar os zeros no final
print("\nWord Index = " , word_index) # imprimindo o dicionário de palavras
print("\nSequences = " , sequences) # imprimindo a sequência de palavras
print("\nPadded Sequences: ", padded) # imprimindo o padding das sequências
```

```python
test_data = [ # novas frases
    'i really love my dog',
    'my dog loves my manatee'
]
```

```python
test_seq = tokenizer.texts_to_sequences(test_data) # criando a sequência de palavras
print("\nTest Sequence = ", test_seq)
```

```python
padded = pad_sequences(test_seq, maxlen=10) #  criando o padding das sequências, com tamanho maximo
print("\nPadded Test Sequence: ", padded)
```
