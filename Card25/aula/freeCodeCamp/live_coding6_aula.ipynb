```python
import spacy  # importa o spaCy

text = "This is a sample number (555) 555-5555."  # define o texto

nlp = spacy.blank("en")  # cria pipeline vazio
ruler = nlp.add_pipe("entity_ruler")  # adiciona o Entity Ruler

patterns = [
    {
        "label": "PHONE_NUMBER",  # define o tipo da entidade
        "pattern": [{"TEXT": {"REGEX": "((\d){3}-(\d){4})"}}]  # procura 3 números, -, e 4 números
    }
]

ruler.add_patterns(patterns)  # adiciona a regra ao pipeline

doc = nlp(text)  # processa o texto

for ent in doc.ents:  # percorre as entidades encontradas
    print(ent.text, ent.label_)  # mostra o texto e o tipo da entidade
```
