```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")
doc = nlp("Britain is a place. Mary is a doctor.") #define doc com uma palavra
```

```python
for ent in doc.ents:
    print (ent.text, ent.label_) #é mostrado as entidades presentes no text e qual é seu tipo
```

```python
from spacy.language import Language
```

```python
@Language.component("remove_gpe") #é registrado a função no pipeline
def remove_gpe(doc):
    original_ents = list(doc.ents) # as entidades originais do texto
    for ent in doc.ents: #verifica se cada entidade do texto se é geopolitica se for, vai remover do doc
        if ent.label_ == "GPE":
            original_ents.remove(ent)
    doc.ents = original_ents
    return (doc)
```

```python
nlp.add_pipe("remove_gpe") #adiciona a função no pipeline
```

```python
nlp.analyze_pipes() #mostra o pipeline completo
```

```python
doc = nlp("Britain is a place. Mary is a doctor.") #define doc com uma palavra
for ent in doc.ents:
    print (ent.text, ent.label_) #é mostrado as entidades presentes no text e qual é seu tipo
```
