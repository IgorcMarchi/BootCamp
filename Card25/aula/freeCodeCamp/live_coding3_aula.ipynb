```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm") # carrega o modelo en_core_web
text = "West Chestertenfieldville was referenced in Mr. Deeds."
```

```python
doc = nlp(text) # cria o doc
```

```python
for ent in doc.ents:
    print(ent.text, ent.label_) # #entidades encontradas no text
```

```python
ruler = nlp.add_pipe("entity_ruler") # adiciona o ruler ao pipeline
```

```python
nlp.analyze_pipes() # analizando a pipeline
```

```python
patterns = [
    {"label" : "GPE", "pattern" : "West Chestertenfieldville"}
]
```

```python
ruler.add_patterns(patterns) # adiciona o padrão ao ruler
```

```python
doc2 = nlp("text")
for ent in doc2.ents:
  print(ent.text, ent.label_)
```

```python
nlp2 = spacy.load("en_core_web_sm") # cria um pipeline sem nada
```

```python
ruler = nlp2.add_pipe("entity_ruler", before='ner') # adiciona o ruler, antes do ner
```

```python
ruler.add_patterns(patterns) # adiciona o padrão ao ruler
```

```python
doc = nlp2(text)
```

```python
for ent in doc.ents:
    print(ent.text, ent.label_) # imprime o texto em ponto e rotulos em pontos
```
