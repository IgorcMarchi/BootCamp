```python
text = nlp("Mike and joy's plane football")
doc2 = nlp(text) #cria outro objeto doc
print(doc2)
```

```python
for token in doc2:
  print(token.text, token.pos_, token.dep_) # mostra o texto, a classe gramatical e a relação sintática
```
