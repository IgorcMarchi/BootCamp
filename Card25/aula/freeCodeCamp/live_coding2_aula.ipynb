```python
your_word = "country" # palavra a ser procurada

ms = nlp.vocab.vectors.most_similar(np.asarray( [nlp.vocab.vectors[nlp.vocab.strings[your_word]]]),n=10)
#busca o indice que corresponde a palavra country no banco vetorial e pega as 10 palavras mais próximas
words = [nlp.vocab.strings[w] for w in ms[0][0]]
distances=ms[2] # distancias
print(words)
```

```python
doc1 = nlp("I like salty fries and hamburgers.") # criando doc1
doc2 = nlp("Fast food tastes very good.") # criando doc2
```

```python
print(doc1, "<->", doc2, doc1.similarity(doc2)) # aqui é apresentado os dois docs e depois o grau de similaridade
```

```python
doc3 = nlp("The Empire State Building is in New York.") # criando doc3
print(doc1, "<->", doc3, doc1.similarity(doc3)) # aqui é apresentado os dois docs e depois o grau de similaridade
```
