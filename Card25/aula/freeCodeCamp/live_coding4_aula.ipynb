```python
import spacy
from spacy.matcher import Matcher
```

```python
nlp = spacy.load("en_core_web_sm") # carrega modelo o modelo en_core_web
```

```python
matcher = Matcher(nlp.vocab) # é usado para procurar padrões em um determinado texto, com base no  vocab escolhido
pattern = [{"LIKE_EMAIL" : True}] # encontra com uma função pronta do spaCy o email presente no texto
matcher.add("EMAIL_ADDRESS", [pattern]) # adiciona a busca no matcher
```

```python
doc = nlp("This is an email address: wmattingly@aol.com") # cria um doc
matches = matcher(doc) # aqui é usado o matches para procurar o email no doc acima
```

```python
print(matches) # mostra os matches encontrados
```

```python
print(nlp.vocab[matches[0][0]].text)
```

```python
print(text)
```

```python
nlp = spacy.load("en_core_web_sm") # carrega o modelo
```

```python
matcher = Matcher(nlp.vocab)
pattern = [{"POS": "PROPN"}]
matcher.add("PROPER_NOUN", [pattern]) # procura um token no qual seja da classe gramatical substantivo próprio
doc = nlp(text)
matches = matcher(doc) #utiliza o matcher para procurar no doc

print (len(matches)) # tamanho dos achados

for match in matches[:10]:
    print (match, doc[match[1]:match[2]]) # mostra o texto do token achad
```

```python
import json
with open ("/content/drive/MyDrive/Colab Notebooks/Card 25/aula/ freeCodeCamp/curso/alice.json") as f: # abrindo o arquivo
  data = json.load(f) # fazendo leitura
```

```python
text = data[0][2][0]
print(text) # mostra a primeira frase do text
```

```python
text = text.replace("`", "'") # troca os caracteres
print (text)
```

```python
matcher = Matcher(nlp.vocab)
pattern = [{"ORTH": "'"},  # pega uma palavra ou conjunto de palavras que estao dentro de aspas, com nenhuma ou mais pontuacoes
           {"IS_ALPHA": True, "OP": "+"}, # + uma ou varias vezes
           {"IS_PUNCT": True, "OP": "*"}, # * zero ou mais vezes
           {"ORTH": "'"}
          ]
matcher.add("PROPER_NOUN", [pattern], greedy="LONGEST") # encontrar a correspondência mais longa possível.

doc = nlp(text)

matches = matcher(doc)
matches.sort(key = lambda x: x[1]) # ordena os matches

print (len(matches)) # tamanho dos encontrados

for match in matches[:10]:
    print (match, doc[match[1]:match[2]])
```

```python
speak_lemmas = ["think", "say"] # verbo da busca
matcher = Matcher(nlp.vocab)
pattern = [{"ORTH": "'"},  # pega uma palavra ou conjunto de palavras que estao dentro de aspas, com nenhuma ou mais pontuacoes
           {"IS_ALPHA": True, "OP": "+"}, # + uma ou varias vezes
           {"IS_PUNCT": True, "OP": "*"}, # * zero ou mais vezes
           {"ORTH": "'"},
           {"POS": "VERB", "LEMMA": {"IN": speak_lemmas}}, # o proximo token precisa ser um verbo e precisa ter uma forma limitada que esta contido ek speak lemmas
           {"POS": "PROPN", "OP": "+"}, # procura o proprio substantiva uma ou varias vezes
           {"ORTH": "'"},  # pega uma palavra ou conjunto de palavras que estao dentro de aspas, com nenhuma ou mais pontuacoes
           {"IS_ALPHA": True, "OP": "+"}, # + uma ou varias vezes
           {"IS_PUNCT": True, "OP": "*"}, # * zero ou mais vezes
           {"ORTH": "'"}
          ]
matcher.add("PROPER_NOUN", [pattern], greedy="LONGEST") # encontrar a correspondência mais longa possível.

doc = nlp(text)

matches = matcher(doc)
matches.sort(key = lambda x: x[1]) # ordena os matches

print (len(matches)) # tamanho dos encontrados

for match in matches[:10]:
    print (match, doc[match[1]:match[2]])
```

```python
matcher = Matcher(nlp.vocab)
pattern1 = [{'ORTH': "'"}, {'IS_ALPHA': True, "OP": "+"}, {'IS_PUNCT': True, "OP": "*"}, {'ORTH': "'"}, {"POS": "VERB", "LEMMA": {"IN": speak_lemmas}}, {"POS": "PROPN", "OP": "+"}, {'ORTH': "'"}, {'IS_ALPHA': True, "OP": "+"}, {'IS_PUNCT': True, "OP": "*"}, {'ORTH': "'"}]
pattern2 = [{'ORTH': "'"}, {'IS_ALPHA': True, "OP": "+"}, {'IS_PUNCT': True, "OP": "*"}, {'ORTH': "'"}, {"POS": "VERB", "LEMMA": {"IN": speak_lemmas}}, {"POS": "PROPN", "OP": "+"}]
pattern3 = [{"POS": "PROPN", "OP": "+"},{"POS": "VERB", "LEMMA": {"IN": speak_lemmas}}, {'ORTH': "'"}, {'IS_ALPHA': True, "OP": "+"}, {'IS_PUNCT': True, "OP": "*"}, {'ORTH': "'"}]
# diferentes combinações de nome proprio e verbbos
# aspas, verbo, nome e aspas
# aspas, verbo, nome se aspas
# nome, verbo aspas

matcher.add("PROPER_NOUNS", [pattern1, pattern2, pattern3], greedy='LONGEST')# maior frase encontarda é  determinando pelas seguintes combinações

for text in data[0][2]:
    text = text.replace("`", "'")
    doc = nlp(text)
    matches = matcher(doc)
    matches.sort(key = lambda x: x[1])
    print (len(matches))
    for match in matches[:10]:
        print (match, doc[match[1]:match[2]])
```
