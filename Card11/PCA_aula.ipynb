```python
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
import pylab as pl
from itertools import cycle

iris = load_iris() # função que carrega os dados da flor iris

numpSamples, numFeatures = iris.data.shape # extraindo dois dados
print (numpSamples) # quantidade de amostras
print (numFeatures) # quantidade de dimensões
print (list(iris.target_names)) # lista de nomes de classicações da flor iris
```

```python
X = iris.data # conjunto de dados
pca = PCA(n_components=2, whiten=True).fit(X) # primeiro definimos a quantidade de dimensões (2) e depois usamos o whiten para normalizar os dados
x_pca = pca.transform(X) # transforma em PCA
```

```python
print(pca.components_) # matriz dos autovetores do plano dos componentes principais
```

```python
print(pca.explained_variance_ratio_) # mostra uma lista da porcentagem da preservação das dimensões
print(sum(pca.explained_variance_ratio_)) # soma a porcentagem de preservação
```

```python
colors = cycle ('rgb') # lista de cores
target_ids = range(len(iris.target_names)) # a cor para cada nome da especie(3)
pl.figure()
for i, c, label in zip(target_ids, colors, iris.target_names): # combinar tudo com os nomes reais de cada especie
    # cada especie tera uma cor correspondente
    pl.scatter(x_pca[iris.target == i,0], x_pca[iris.target == i,1], c=c, label=label)
pl.legend() # cria a legenda
pl.show() # exibe a iamgem
```
