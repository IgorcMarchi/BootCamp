```python
import numpy as np
from sklearn import model_selection # corrigindo porque mudou o cross paraa model_selection
from sklearn import datasets
from sklearn import svm

iris = datasets.load_iris() # carregando datasets da iris
```

```python
# dividindo o dataset em treino e teste
x_train, x_test, y_train, y_test = model_selection.train_test_split(iris.data, iris.target, test_size=0.4, random_state=0)
clf = svm.SVC(kernel='linear', C=1).fit(x_train, y_train) # criando um modelo de previsao  para prever as classes do iris atraves dos treinamentos
clf.score(x_test, y_test) # aveliando o modelo do test comn dados
```

```python
# fazendo a validação cruzada dando o modelo, o dataset e os valores reais, alem do numero de folds onde sera executado 5 vezes
scores = model_selection.cross_val_score(clf, iris.data, iris.target, cv=5)

print(scores) # lista da precisão dos folds
print(scores.mean()) # media da precisao
```

```python
# testando o projeto com maior dimensão(Kernel poly)
clf = svm.SVC(kernel='poly', C=1).fit(x_train, y_train)
scores = model_selection.cross_val_score(clf, iris.data, iris.target, cv=5)
print(scores)
print(scores.mean())
```

```python
# outro modelo com kernel poly
clf = svm.SVC(kernel='poly', C=1).fit(x_train, y_train)

# medida com os dados de teste
clf.score(x_test, y_test)
```

## Activity
The "poly" kernel for SVC actually has another attribute for the number of degrees of the polynomial used, which defaults to 3. For example, svm.SVC(kernel='poly', degree=3, C=1)

```python
# trocando o numero de graus do kernel
clf = svm.SVC(kernel='poly', degree=2, C=5).fit(x_train, y_train)
clf.score(x_test, y_test)
```

```python
clf = svm.SVC(kernel='poly', degree=3, C=5).fit(x_train, y_train)
clf.score(x_test, y_test)
```

```python
clf = svm.SVC(kernel='poly', degree=4, C=5).fit(x_train, y_train)
clf.score(x_test, y_test)
```
