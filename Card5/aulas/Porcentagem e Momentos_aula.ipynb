```python
import numpy as np 
import matplotlib.pyplot as plt
vals = np.random.normal(0, 0.5, 10000) # gera dados com média 0 e desvio 0.5

plt.hist(vals, 50) # cria histograma com 50
plt.show()
```

```python
np.percentile(vals, 50) # calcula o valor do percentual de 50
```

```python
np.percentile(vals, 90) # calcula o valor do percentual de 90
```

```python
np.percentile(vals, 20) # calcula o valor do percentual de 20
```

```python
np.percentile(vals, 99) # calcula o valor do percentual de 99
```

## Primeiro momento

```python
np.mean(vals) # media
```

## Segundo momento

```python
np.var(vals) # Variancia elevado a 2
```

## Terceiro momento

```python
import scipy.stats as sp
sp.skew(vals) # assimetria
```

## Quarto momento

```python
sp.kurtosis(vals) #distribuicao normal
```
