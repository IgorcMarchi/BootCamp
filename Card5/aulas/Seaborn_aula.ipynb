```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt

# Lê o arquivo CSV a partir da URL e cria um DataFrame
df = pd.read_csv("http://media.sundog-soft.com/SelfDriving/FuelEfficiency.csv")

# Conta quantas vezes cada quantidade de marchas aparece
gear_counts = df['# Gears'].value_counts()
gear_counts.plot(kind='bar') # Cria um gráfico de barras com as contagens
plt.show() # exibe
```

```python
import seaborn as sns
sns.set() # Aplica o estilo padrão
```

```python
gear_counts.plot(kind='bar') # cria um grafico a partir da quantidade de 'bar'
plt.show()
```

```python
df.head() # Mostra as primeiras 5 linhas do DataFrame
```

```python
sns.displot(df['CombMPG']) # Cria um histograma dos valores de consumo combinado
plt.show()
```

```python
df2 = df[['Cylinders', 'CityMPG', 'HwyMPG', 'CombMPG']] # Cria um novo DataFrame apenas com as colunas selecionadas
df2.head() # Mostra as primeiras 5 linhas do novo DataFrame
```

```python
sns.pairplot(df2, hue='Cylinders', height=2.5); # Cria gráficos de dispersão e histogramas para cada par de variáveis
# usando a coluna 'Cylinders' para diferenciar as cores
plt.show()
```

```python
# Cria um gráfico de dispersão
sns.scatterplot(x="Eng Displ", y="CombMPG", data=df)
plt.show()
```

```python
# Cria um jointplot mostrando a relações e consumo combinado
sns.jointplot(x="Eng Displ", y="CombMPG", data=df)
plt.show()
```

```python
# Cria um gráfico de dispersão com linha de regressão linear
sns.lmplot(x="Eng Displ", y="CombMPG", data=df)
plt.show()
```

```python
sns.set(rc={'figure.figsize':(15,5)}) # Define o tamanho padrão das figuras
ax=sns.boxplot(x='Mfr Name', y='CombMPG', data=df) # Cria um boxplot do consumo combinado de x e y
ax.set_xticklabels(ax.get_xticklabels(),rotation=45) # Rotaciona em 45 graus
```

```python
plt.show() # exibe grafico
```

```python
ax=sns.swarmplot(x='Mfr Name', y='CombMPG', data=df)
ax.set_xticklabels(ax.get_xticklabels(),rotation=45)
```

```python
plt.show() # exibe grafico
```

```python
ax=sns.countplot(x='Mfr Name', data=df)
ax.set_xticklabels(ax.get_xticklabels(),rotation=45)
```

```python
plt.show()
```

```python
# Cria uma tabela dinâmica com a média do consumo combinado e número de cilindros como índice e cilindrada do motor como colunas
df2 = df.pivot_table(index='Cylinders', columns='Eng Displ', values='CombMPG', aggfunc='mean') # Novo DataFrame reorganizando o original
sns.heatmap(df2) # Cria um mapa de calor 
plt.show() # exibe
```
