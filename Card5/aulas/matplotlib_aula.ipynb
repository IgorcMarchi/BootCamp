```python
plt.xkcd()  # ativa estilo XKCD

fig = plt.figure()  # cria a figura
ax = fig.add_subplot(1, 1, 1)  # adiciona subplot
ax.spines['right'].set_color('none')  # remove borda direita
ax.spines['top'].set_color('none')  # remove borda superior

plt.xticks([])  # remove marcações do eixo X
plt.yticks([])  # remove marcações do eixo Y
ax.set_ylim([-30, 10])  # limites do eixo Y

data = np.ones(100)  # cria vetor de dados
data[70:] -= np.arange(30)  # queda a partir do índice 70

plt.annotate(
    'Quando tem aula de algebra', 
    xy=(70, 1), arrowprops=dict(arrowstyle='->'), xytext=(15, -10))  # seta e posição

plt.plot(data)  # plota os dados

plt.xlabel('Tristeza')  # rótulo do eixo X
plt.ylabel('Tempo')  # rótulo do eixo Y

```

```python
plt.rcdefaults() # define o estilo rcdefaults

values = [12, 55, 4, 32, 14] # define os valores
colors = ['r', 'g', 'b', 'c', 'm'] # define as cores
explode = [0, 0, 0.2, 0, 0] # define o tamanho expansivo
labels = ['India', 'EUA', 'Rusia', 'China', 'Europe']  # fefine os nomes (países)
plt.pie(values, colors= colors, labels= labels, explode= explode) # cria gráfico de pizza
plt.title('Students Location') # adiciona o titulo
plt.show()
```

```python
uniformSkewed = np.random.rand(100) * 100 - 40  # dados uniformes deslocados
high_outliers = np.random.rand(10) * 50 + 100  # outliers altos
low_outliers = np.random.rand(10) * -50 - 100  # outliers baixos
data = np.concatenate((uniformSkewed, high_outliers, low_outliers))  # junta os dados
plt.boxplot(data)  # cria boxplot
plt.show()  # exibe o gráfico
```
