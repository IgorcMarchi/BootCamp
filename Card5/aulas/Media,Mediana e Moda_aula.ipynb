```python
import numpy as np
incomes = np.random.normal(27000, 15000, 10000)  # gera dados de renda
np.mean(incomes)  # calcula a média
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
plt.hist(incomes, 50) # cria histograma
plt.show() # exibe
```

```python
np.median(incomes)  # calcula a mediana
```

```python
incomes = np.append(incomes, [1000000000]) # adiciona um valor extremo
```

```python
np.median(incomes) # calcula novamentge a mediana
```

```python
np.mean(incomes) # calcula a media
```

```python
ages = np.random.randint(18, high=90, size=500) # cria um array de de idades de 18 a 90, com 500 numeros aleatorios
ages
```

```python
from scipy import stats
stats.mode(ages) # mostra a quantidade do numero 'moda' e qual foi o numero da 'moda'
```
