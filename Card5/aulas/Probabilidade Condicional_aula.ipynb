```python
from numpy import random
random.seed(0)  # fixa a semente para resultados reproduzíveis

totals = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}  # total de pessoas por faixa etária
purchases = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}  # total de compras por faixa
totalPurchases = 0  # contador total de compras

for _ in range(100000):  # simula 100 mil pessoas
    ageDecade = random.choice([20, 30, 40, 50, 60, 70])  # escolhe faixa etária
    purchaseProbability = float(ageDecade) / 100.0  # probabilidade de compra
    totals[ageDecade] += 1  # incrementa total da faixa
    if random.random() < purchaseProbability:  # verifica se houve compra
        totalPurchases += 1  # incrementa compras totais
        purchases[ageDecade] += 1  # incrementa compras da faixa
```

```python
totals # mostra os valores total de pessoas por faixas etária
```

```python
purchases # mostra o total de compras por faixas
```

```python
totalPurchases #  mostra o total de compras contados
```

```python
PEF = float(purchases[30]) / float(totals[30]) # compra e idade = 30
print('P(purchase | 30s): ' + str(PEF)) # exibe a probabilidade
```

```python
PF = float(totals[30]) / 100000.0 # probabilidade de alguém estar na faixa dos 30
print("P(30's): " +  str(PF)) # mosta a probabilidad
```

```python
PE = float(totalPurchases) / 100000.0 # probabilidade total de compra
print("P(Purchase):" + str(PE)) # exibe a probabilidade
```

```python
print("P(30's, Purchase)" + str(float(purchases[30]) / 100000.0)) # probabilidade conjunta (30 e compra
```

```python
print("P(30's)P(Purchase)" + str(PE * PF)) # produto das probabilidades P(30) e P(compra)
```
