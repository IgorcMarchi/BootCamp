```python
import matplotlib.pyplot as plt

import numpy as np
from pylab import *


def de_mean(x):
    xmean = mean(x) # Calcula a média do vetor
    return [xi - xmean for xi in x] # Subtrai a média de cada elemento


def covariance(x, y):
    n = len(x) # Número de elementos
    return dot(de_mean(x), de_mean(y)) / (n-1) # Calcula a covariância entre x e y

# Gera dados aleatórios
pageSpeeds = np.random.normal(3.0, 1.0, 1000)
purchaseAmount = np.random.normal(50.0, 10.0, 1000)

# Cria o gráfico de dispersão entre as variáveis
scatter(pageSpeeds, purchaseAmount)

# Calcula a covariância entre pageSpeeds e purchaseAmount
covariance (pageSpeeds, purchaseAmount)
```

```python
# Gera valores de compra e os divide pela velocidade da página
purchaseAmount = np.random.normal(50.0, 10.0, 1000) / pageSpeeds


# Plota a relação entre velocidade da página e valor de compra
scatter(pageSpeeds, purchaseAmount)
covariance(pageSpeeds, purchaseAmount) # Calcula a covariância entre as duas variáveis
```

```python
def correlation(x, y):
    stddevx = x.std() # Desvio padrão de x
    stddevy = y.std() # Desvio padrão de y
    return covariance(x,y) / stddevx / stddevy  

# Calcula a correlação
correlation(pageSpeeds, purchaseAmount)
```

```python
np.corrcoef(pageSpeeds, purchaseAmount) # Calcula a matriz de correlação
```

```python
# Define o valor de compra
purchaseAmount = 100 - pageSpeeds * 3

# Plota o valor
scatter(pageSpeeds, purchaseAmount)

# Calcula a correlação
correlation (pageSpeeds, purchaseAmount)
```
