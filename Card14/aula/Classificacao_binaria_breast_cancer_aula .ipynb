```python
#previsores
X = pd.read_csv('/content/entradas_breast.csv')
X
```

```python
# biblioteca
import sklearn
from sklearn.model_selection import train_test_split
```

```python
x_treinamento, x_teste, y_treinamento, y_teste = train_test_split(X, y, test_size = 0.25) # 25 % dos dados para testar e os 75% para treinar
```

```python
# biblioteca
import tensorflow as tf
from tensorflow.keras.models import Sequential
```

```python
tf.__version__ # versão
```

```python
(30 + 1) / 2 # calculo de neuronios da camada oculta
```

```python
rede_neural = Sequential([
    tf.keras.layers.InputLayer(shape = (30,)),
    tf.keras.layers.Dense(units = 16, activation= 'relu', kernel_initializer = 'random_uniform'),
    tf.keras.layers.Dense(units = 16, activation= 'relu', kernel_initializer = 'random_uniform'), # foi acrescentado depois
    tf.keras.layers.Dense(units = 1, activation= 'sigmoid')
]) # rede neural para duas camadas oculta
```

```python
30 * 16, 496 - 480 # Bias
```

```python
496 + 17 # parametros totais
```

```python
rede_neural.summary()
```

```python
otimizador = tf.keras.optimizers.Adam(learning_rate = 0.001, clipvalue = 0.5) # Criando o otimizador
```

```python
rede_neural.compile(optimizer = otimizador, loss = 'binary_crossentropy', metrics = ['binary_accuracy']) # definindo os parametros
```

```python
426 / 10 # calculo do batch size
```

```python
rede_neural.fit(x_treinamento, y_treinamento, batch_size = 10, epochs = 100) # processamento
```

```python
previsoes = rede_neural.predict(x_teste) # celula da variavel de previsoes
previsoes = previsoes > 0.5
```
