## ReLu

```python
def relu_function(soma): # Função onde qualquer valor maior ou igual a zero retorna o proprio valor
#Porém caso seja menor que zero então retorna o valor 0
  if soma>= 0:
    return soma
  return 0
```

```python
relu_function(-20) # teste
```

```python
relu_function(0) # teste
```

```python
relu_function(10) # teste
```

```python
relu_function(100) # teste
```

## SoftMax

```python
def softmax_function(x): # a função retorna a probabilidade para mais de duas classes
  ex = np.exp(x)
  return ex / ex.sum()
```

```python
softmax_function([7, 2, 1.3]) # teste
```

```python
softmax_function([2, 2, 5.3]) # teste
```
