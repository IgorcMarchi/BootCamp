```python
# bibliotecas
import pandas as pd
import numpy as np
import tensorflow as tf
import sklearn
import scikeras
```

```python
from scikeras.wrappers import KerasClassifier
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras import utils as np_utils
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from tensorflow.keras import backend as K
from sklearn.model_selection import cross_val_score
```

```python
base = pd.read_csv('/content/iris.csv') # leitura do csv dos dados
X = base.iloc[:, 0:4] # todas as linhas e pegas as colunas 0,1,2 e 3
y = base.iloc[:, 4].values # mostras os valores das linhas apenas da coluna 4
```

```python
label_encoder = LabelEncoder() # transforma as variaveis string em numeros(0,1,2)
y = label_encoder.fit_transform(y)
y = np_utils.to_categorical(y)
```

```python
def criar_rede():
  rede_neural = Sequential([
      tf.keras.layers.InputLayer(input_shape=(4,)),
      tf.keras.layers.Dense(units=4, activation='relu'), # camada oculta
      tf.keras.layers.Dense(units=4, activation='relu'),
      tf.keras.layers.Dense(units=3, activation='softmax') # camada de saida, softmax pq tem o uso de mais de 2 classes
  ]) # rede neural onde cria uma sequencia de camada

  rede_neural.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy']) # compila a rede neural
  return rede_neural
```

```python
rede_neural = KerasClassifier(model=criar_rede, epochs=250, batch_size=10)
```

```python
resultados = cross_val_score(estimator=rede_neural, X=X, y=y, cv=10, scoring='accuracy')
```
