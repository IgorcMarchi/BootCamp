```python
# bibliotecas
import pandas as pd
import numpy as np
import tensorflow as tf
import sklearn
```

```python
pd.__version__, np.__version__, tf.__version__, sklearn.__version__
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras import utils as np_utils
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
```

```python
base = pd.read_csv('/content/iris.csv') # leitura do csv dos dados
base
```

```python
X = base.iloc[:, 0:4] # todas as linhas e pegas as colunas 0,1,2 e 3
X
```

```python
y = base.iloc[:, 4].values # mostras os valores das linhas apenas da coluna 4
y
```

```python
label_encoder = LabelEncoder() # transforma as variaveis string em numeros(0,1,2)
y = label_encoder.fit_transform(y)
y
```

```python
np.shape(y) # quantidade
```

```python
X_treinamento, X_teste, y_treinamento, y_teste = train_test_split(X, y, test_size=0.25) # separa os dados em treinamento e teste
```

```python
X_treinamento.shape, X_teste.shape # quantidade de registros para treinamento e teste de X
```

```python
y_treinamento.shape, y_teste.shape # quantidade de registros para treinamento e teste de y
```

```python
(4 + 3) / 2 # quantidade de neuronios na camada oculta
```

```python
rede_neural = Sequential([
    tf.keras.layers.InputLayer(input_shape=(4,)),
    tf.keras.layers.Dense(units=4, activation='relu'), # camada oculta
    tf.keras.layers.Dense(units=4, activation='relu'),
    tf.keras.layers.Dense(units=3, activation='softmax') # camada de saida, softmax pq tem o uso de mais de 2 classes
]) # rede neural onde cria uma sequencia de camada
```

```python
rede_neural.summary()
```

```python
rede_neural.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['categorical_accuracy']) # compilar a rede
```

```python
rede_neural.fit(X_treinamento, y_treinamento, batch_size=10, epochs=1000) # treinar a rede
```

```python
rede_neural.evaluate(X_teste, y_teste) # avaliar a rede
```

```python
previsoes = rede_neural.predict(X_teste) # previsoes utilizando a variavel X_teste
previsoes
```

```python
previsoes = previsoes > 0.5 # facilitar o entendimento dos valores dividindo em True e False
print(previsoes)
```

```python
y_teste2 = [int(np.argmax(t)) for t in y_teste]
print(y_teste2) # representação no formato original das classes
```

```python
previsoes2 = [int(np.argmax(t)) for t in previsoes]
print(previsoes2) # representação no formato original das classes
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_teste2, previsoes2) # acuracia do modelo
```

```python
# 0 setosa, 1 versicolor, 2 virginica
```

```python
confusion_matrix(y_teste2, previsoes2)
```
