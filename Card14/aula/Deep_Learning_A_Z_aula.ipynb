## Step Function

```python
def step_function(soma): #Função que se o numero for maior ou igual a 1 ele retorna 1 se menor ele volta 0
  if soma >= 1:
    return 1
  return 0
```

```python
step_function(-10) # teste
```

```python
step_function(10) # teste
```

```python
step_function(1) # teste
```

## Sigmoid

```python
def sigmoid_function(soma): # função que calcula a formula sigmoid
  return 1 / (1 + np.exp(-soma))
```

```python
sigmoid_function(-10) # teste
```

```python
sigmoid_function(10) # teste
```

```python
sigmoid_function(1) # teste
```

## Hyperbolic

```python
def tahn_function(soma): # Função da tangente hiperbólica
  return (np.exp(soma) - np.exp(-soma)) / (np.exp(soma) + np.exp(-soma))
```

```python
tahn_function(-10) # teste
```

```python
tahn_function(10) # teste
```

```python
tahn_function(1) # teste
```
