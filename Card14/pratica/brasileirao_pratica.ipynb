```python
# Manipulação de dados
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split # Divisão dos dados em treino e teste
from sklearn.preprocessing import StandardScaler # Normalização dos dados

# TensorFlow/Keras
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Dropout

from sklearn.preprocessing import LabelEncoder # Transformação de Texto em Números

from sklearn.metrics import accuracy_score, mean_absolute_error # Métricas de avaliação

# Avaliação do modelo
from sklearn.metrics import accuracy_score, mean_absolute_error
```

```python
df = pd.read_csv("/content/brasileirao.csv") # csv
df
```

```python
df.shape # quantidade de linhas e colunas
```

```python
df.columns # nomes das colunas
```

```python
df.info() # informações gerais
```

```python
# Remove colunas criadas automaticamente pelo Pandas
df.drop(columns=["Unnamed: 0","Unnamed: 0.1", "Unnamed: 0.2", "ID","Data"],inplace=True)

# Verifica o resultado
df.head()
```

```python
# Transformação de Texto em Números

encoder_casa = LabelEncoder()
encoder_fora = LabelEncoder()
encoder_primeiro = LabelEncoder()

df["Casa"] = encoder_casa.fit_transform(df["Casa"])
df["Fora"] = encoder_fora.fit_transform(df["Fora"])

# Trata valores nulos da coluna PrimeiroGol
df["PrimeiroGol"] = df["PrimeiroGol"].fillna("Nenhum")

df["PrimeiroGol"] = encoder_primeiro.fit_transform(df["PrimeiroGol"])
```

```python
# 1 = vitória do mandante
# 0 = empate ou derrota
df["Vitoria"] = (df["Gol_Casa"] > df["Gol_Fora"]).astype(int) # criando coluna de vitoria
df[["Gol_Casa","Gol_Fora","Vitoria"]].head()
```

```python
# Definição das Entradas (X) e Saídas (y)
X = df.drop(columns=["Gol_Casa","Gol_Fora","Vitoria"]).values

y_class = df["Vitoria"].values

y_gols_casa = df["Gol_Casa"].values

y_gols_fora = df["Gol_Fora"].values
```

```python
# treinamento de 80 % e 20 % de teste
X_train, X_test, y_c_train, y_c_test, y_gc_train, y_gc_test, y_gf_train, y_gf_test = train_test_split(X, y_class, y_gols_casa, y_gols_fora, test_size=0.2, random_state=42)
```

```python
# Normalizando os Dados
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
input_layer = Input(shape=(X.shape[1],)) # Camada de entrada
shared = Dense(64, activation="relu")(input_layer) # Camadas ocultas compartilhadas

shared = Dropout(0.3)(shared)
shared = Dense(32, activation="relu")(shared)

out_class = Dense(1, activation="sigmoid", name="vitoria")(shared) # Saída de classificação

out_gols_casa = Dense(1, activation="linear", name="gols_casa")(shared) # Saída de regressão
out_gols_fora = Dense(1, activation="linear", name="gols_fora")(shared) # Saída de regressão
```

```python
model = Model(inputs=input_layer,outputs=[out_class,out_gols_casa,out_gols_fora]) # montagem do modelo
```

```python
model.compile(optimizer="adam", # compilacao de rede
    loss={
        "vitoria":"binary_crossentropy",
        "gols_casa":"mse",
        "gols_fora":"mse"
    },
    metrics={
        "vitoria":"accuracy",
        "gols_casa":"mae",
        "gols_fora":"mae"})
```

```python
historico = model.fit(X_train,[y_c_train, y_gc_train, y_gf_train],epochs=50,batch_size=32,verbose=1) # treinamento do modelo
```

```python
previsoes = model.predict(X_test) # previsoes
```

```python
#Avaliacao de classificação
vitoria_pred = (previsoes[0] > 0.5).astype(int)

acuracia = accuracy_score(y_c_test, vitoria_pred)
print("Acurácia:", acuracia)
```

```python
erro_gols_casa = mean_absolute_error( y_gc_test, previsoes[1]) # Erro medio dos gols de casa
print("MAE Gols Casa:", erro_gols_casa)
```

```python
erro_gols_fora = mean_absolute_error(y_gf_test, previsoes[2]) #  Erro medio dos gols visitantes
print("MAE Gols Fora:", erro_gols_fora)
```
