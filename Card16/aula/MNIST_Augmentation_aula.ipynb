```python
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import InputLayer, Dense, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import utils as np_utils
```

```python
(X_treinamento, y_treinamento), (X_teste, y_teste) = mnist.load_data() # carrega
```

```python
X_treinamento = X_treinamento.reshape((X_treinamento.shape[0], 28, 28, 1))
X_teste = X_teste.reshape((X_teste.shape[0], 28, 28, 1))
X_treinamento = X_treinamento.astype('float32')
X_teste = X_teste.astype('float32')
X_treinamento /= 255
X_teste /= 255
y_treinamento = np_utils.to_categorical(y_treinamento, 10)
y_teste = np_utils.to_categorical(y_teste, 10)
```

```python
classificador = Sequential()
classificador.add(InputLayer(input_shape=(28, 28, 1))) # camada de entrada
classificador.add(Conv2D(32, (3, 3), activation='relu')) # camada convolucional
classificador.add(MaxPooling2D(pool_size=(2, 2))) # camada de pooling
classificador.add(Flatten()) # camada de flattening
classificador.add(Dense(units=128, activation='relu')) # camada oculta
classificador.add(Dense(units=10, activation='softmax')) # camada de saída
classificador.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
gerador_treinamento= ImageDataGenerator(rotation_range=7, horizontal_flip=True, shear_range=0.2, height_shift_range=0.07, zoom_range=0.2) # Criando a variavel de gerador de treimenti configurando ele
```

```python
gerador_teste = ImageDataGenerator() # Criando a variavel de gerador de teste mantendo as proprias imagens sem alterações
```

```python
base_treinamento = gerador_treinamento.flow(X_treinamento, y_treinamento, batch_size=128) # Gerando a base de treinamento
```

```python
base_teste = gerador_teste.flow(X_teste, y_teste, batch_size=128) # Gerando a base de teste
```

```python
classificador.fit(base_treinamento, epochs=5, validation_data=base_teste)
```
