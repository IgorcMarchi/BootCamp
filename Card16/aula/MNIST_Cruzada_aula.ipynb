```python
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import InputLayer, Dense, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras import utils as np_utils
from sklearn.model_selection import StratifiedKFold
```

```python
(X, y), (X_teste, y_teste) = mnist.load_data() # Carregamento dos dados
X = X.reshape(X.shape[0], 28, 28, 1)
X = X.astype('float32') #convertendo para float 32
X /= 225 # para que os pixels fiquem na faixa entre 0 e 1
y = np_utils.to_categorical(y, 10) # convertendo para camada de saida da rede neural
```

```python
np.zeros(shape = (y.shape[0], 1))
```

```python
seed = 5
np.random.seed(seed)
```

```python
kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=seed) # definindo a variavel kfold
```

```python
resultados = []
for indice_treinamento, indice_teste in kfold.split(X, np.zeros(shape = (y.shape[0], 1))):
  print('Indice Treinamento: ', indice_treinamento, 'Indice Teste: ', indice_teste)
  classificador = Sequential()

  classificador.add(InputLayer(shape=(28, 28, 1)))   #Camada de convolução
  classificador.add(Conv2D(32, (3, 3), activation='relu'))
  classificador.add(MaxPooling2D(pool_size=(2,2)))
  classificador.add(Flatten())

  classificador.add(Dense(units=128, activation='relu')) # Camada oculta

  classificador.add(Dense(units=10, activation='softmax'))
  classificador.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy']) # Camada saída
  classificador.fit(X[indice_treinamento], y[indice_treinamento], batch_size = 128, epochs = 5) #usando dados de treino com o índice de treinamento
  precisao = classificador.evaluate(X[indice_teste], y[indice_teste]) # usando dados de teste com o índice de teste
  resultados.append(precisao[1])
```
