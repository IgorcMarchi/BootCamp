```python
try:
  %tensorflow_version 2.0
except Exception:
  pass

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

print(tf.__version__)
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
sentences = [
    'I like eggs and ham.',
    'I love chocolate and bunnies',
    'I hate onions'
]
```

```python
MAX_VOCAB_SIZE = 20000 # tamanho máximo
tokenizer = Tokenizer(num_words=MAX_VOCAB_SIZE) # cria um token com o maximo
tokenizer.fit_on_texts(sentences) # apaga os espaços
sequences = tokenizer.texts_to_sequences(sentences) # converte
```

```python
print(sequences)
```

```python
tokenizer.word_index
```

```python
data = pad_sequences(sequences) # padding das sequências
print(data)
```

```python
MAX_SEQUENCE_LENGTH = 5 # limita em 5 palavras
data = pad_sequences(sequences, maxlen=MAX_SEQUENCE_LENGTH)
print(data)
```

```python
data = pad_sequences(sequences, maxlen=MAX_SEQUENCE_LENGTH, padding='post') # padding no final
print(data)
```

```python
data = pad_sequences(sequences, maxlen=4) # tamanho da sequência reduzido
print(data)
```

```python
data = pad_sequences(sequences, maxlen=4, truncating='post')
print(data)
```
