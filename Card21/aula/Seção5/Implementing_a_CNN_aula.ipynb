```python
import torch
device = 'cuda' if torch.cuda.is_available() else 'cpu'

# tensores
x_train = torch.tensor([[[[4,5,6,7], [5,6,7,8], [8,9,10,11], [4,6,7,8]]], [[[-4, 5,6, -7], [5, -6, 7, 8], [-8, 9, -10, 11], [-4, -6, -7, -8]]]]).float().to(device)
x_train.div_(8) # dividi o tensor em 8
y_train = torch.tensor([0,1]).float().to(device) # tensor com 2 formatos 0 e 1
```

```python
import torch
import torch.nn as nn
from torch.optim import Adam

def get_model(): # cria uma rede neural utilizando o Sequential
  model = nn.Sequential(
      nn.Conv2d(in_channels=1, out_channels=1, kernel_size=3), # camada convolucional
      nn.MaxPool2d(kernel_size=2), # camada booleana
      nn.ReLU(), # funçao de ativacao
      nn.Flatten(), # camada de achatamento
      nn.Linear(1,1), # camada de saida
      nn.Sigmoid() # funcao de ativacao
  ).to(device)

  loss_fn = nn.BCELoss() # funcao de perda
  optimizer = Adam(model.parameters(), lr=0.01) # otimizador

  return model, loss_fn, optimizer
```

```python
model, criterion, optimizer = get_model()
```

```python
import torch

def train_batch(x,y, model, optimizer, loss_fn): # treina um unico lote de dados
  model.train() # defini como modelo de treinamento
  optimizer.zero_grad() # inicia gradiante em 0
  prediction = model(x)
  batch_loss = loss_fn(prediction.squeeze(), y.squeeze())
  batch_loss.backward()
  optimizer.step()
  return batch_loss.item()
```

```python
from torch.utils.data import TensorDataset, Dataset, DataLoader

trn_dl = DataLoader(TensorDataset(x_train, y_train)) # cria um dataloader
```

```python
import torch
for epoch in range(2000): # carrega 2000 de epoca
  for ix, batch in enumerate(trn_dl):
    x, y = batch
    x = x.to(device)
    y = y.to(device)
    batch_loss = train_batch(x, y, model, optimizer, criterion)
    # durante o treinamento é consultado o carregador de dados, recuperando cada lote de dados e depois treina utilizando a fucao train_batch
```
