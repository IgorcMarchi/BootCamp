```python
import math

# (1/(1+e^(-x)))
def sigmoid(x):
  return 1 / (1 + math.exp(-x))
```

## **ReLU**

```python
import torch
# Implementa a função de Ativação ReLU
class ReLU(torch.nn.Module):
  def __init__(self):
    super().__init__()

  def forward(self, x):
    return torch.max(torch.zeros_like(x), x)
```

## LOSS Functions

## Mean-squared error

```python
import numpy as np
# Essa função recebe dois valores onde um representa os valores verdadeiros e o outro os valores previstos, a função calcula a média dos quadrados das diferenças entre y_true e y_pred
def mse_loss(y_true, y_pred):
  return np.mean((y_true-y_pred)**2)
```

## Cross entropy Loss

#L = -(1/N) * sum(y_true*log(y_pred)+(1-y_true)*log(1-y_pred))

```python
import numpy as np
# Essa função recebe 2 argumentos, que representa os valores verdadeiros e o outro os valores previstos
# A função calcula o logaritomo negativo da verossimilhança da distribuição de probabilidade prevista de y_pred para o rotulo verdadeiro(y_true)
def cross_entropy_loss(y_true, y_pred):
  epsilon = 1e-15
  y_pred = np.clip(y_pred, epsilon, 1-epsilon) # To avoid zero division
  return -np.mean(y_true*np.log(y_pred)+(1-y_true)*np.log(1-y_pred))
```
