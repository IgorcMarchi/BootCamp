```python
import torch
from torchviz import make_dot
```

## Formulação do problema: prever o padrão usando dados de treinamento

```python
# O y é a soma dos elementos de x. Ex: (3 + 4) = [7]
x = [[1,2],[3,4],[5,6],[7,8]]
y = [[3],[7],[11],[15]]
```

```python
#Converter listas em tensores para poder usá-los com PyTorch
X = torch.tensor(x).float()
Y = torch.tensor(y).float()
```

## Para usar a GPU no Google Colab, você precisa alterar o tipo de tempo de execução para GPU.

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(device)
```

```python
X = X.to(device)
Y = Y.to(device)
```

## Para criar um modelo PyTorch personalizado, precisamos herdar da classe nn.module

```python
# definimos a rede neural usando o modulo nn
import torch.nn as nn
```

```python
# Arquitetura da rede neural usando o modulo nn
class MyNeuralNet(nn.Module):
  def __init__(self):
    #When we call the super._initi_() method we ensure we are inhertiting
    super().__init__()
    self.layer = nn.Linear(2, 8) # A linear layer
    self.activation = nn.ReLU() #activation function
    self.Layer2 = nn.Linear(8, 1)

  #When we pass something through the model object, it calls the forward method
  def forward(self, x):
      x = self.layer(x)
      x = self.activation(x)
      x = self.Layer2(x)
      return x
```

```python
model = MyNeuralNet().to(device) # instancia da rede neural e move para o dispositivo
```

# Vizualize a arquitetura do modelo

```python
make_dot(model(X), params=dict(model.named_parameters())).render("dense_network", format="png") # cria o modelo da arquitetura em png
```
