```python
from torch.utils.data import Dataset, DataLoader
import torch
import torch.nn as nn
```

## Define the same dataset as in the previous lessons

```python
# O y é a soma dos elementos de x. Ex: (3 + 4) = [7]
x = [[1,2],[3,4],[5,6],[7,8]]
y = [[3],[7],[11],[15]]
```

```python
#Converter listas em tensores para poder usá-los com PyTorch
X = torch.tensor(x).float()
Y = torch.tensor(y).float()
```

```python
device = 'cuda' if torch.cuda.is_available() else 'cpu'
X = X.to(device)
Y = Y.to(device)
```

```python
class MyDataset(Dataset):
    def __init__(self, x, y):
        self.x = torch.tensor(x).float()
        self.y = torch.tensor(y).float()

    def __len__(self):
        return len(self.x)

    def __getitem__(self, ix):
        return self.x[ix], self.y[ix]
```

```python
ds = MyDataset(x, y)
```

```python
# Notice Batch size
dl = DataLoader(ds, batch_size=2, shuffle=True)
```

```python
# To load the data we loop through it
for x,y in dl:
  print(x,y)
```

## Using the DataLoader object in the training code

```python
import torch
import torch.nn as nn
from torch.optim import SGD
```

```python
class MyNeuralNet(nn.Module):
  def __init__(self):
    super().__init__()
    self.Layer1 = nn.Linear(2,8)
    self.activation = nn.ReLU()
    self.Layer2 = nn.Linear(8,1)

  def forward(self,x):
    x = self.Layer1(x)
    x = self.activation(x)
    x = self.Layer2(x)
    return x
```

```python
model = MyNeuralNet()
loss_func = nn.MSELoss()
opt = SGD(model.parameters(), lr = 0.001)
```

```python
losses = []
for _ in range(50): # for de 50 épocas
  for data in dl:
    opt.zero_grad()
    x1,y1 = data
    loss_value = loss_func(model(x1),y1)
    loss_value.backward()

    opt.step()
    losses.append(loss_value.detach().numpy())
```

## By using a dataloader obect we are able to train the model much faster, as batch processing is taking place
