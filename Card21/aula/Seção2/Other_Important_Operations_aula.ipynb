## Multiplicação de matrizes
Se A e B são duas matrizes, podemos multiplicá-las apenas se tiverem as dimensões A (n x m) e B (m x z), o que resulta em uma matriz de dimensões C (n x z).


```python
x = torch.tensor([[1,2,3,4], [5,6,7,8]]) # shape (2,4)
y = torch.tensor([[1,2,3],
                  [2,3,4],
                  [4,5,6],
                  [7,8,9]]) # shape (4,3)
```

```python
# é usado o matmul para multiplicar os tensors
torch.matmul(x,y) # shape (2,3)
```

## Concatenação

```python
# é dobrado o numero de linhas
x = torch.randn(1,4,5)
z = torch.cat([x,x], axis=1)
print('concatenated axis 1:', x.shape, z.shape)
```

## Permutação

```python
x = torch.randn(3,20,10)
z = x.permute(2,0,1) # trocamos a posição dos valores
print('Permute dimensions', x.shape, z.shape)
```
