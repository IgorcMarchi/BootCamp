## Remodelar
Existem várias maneiras de remodelar um tensor.

## 1) Visualizar

```python
# 1-) Visualizar
x
```

```python
x.shape
```

```python
x.view(4,2) # é remodelado os valores das linhas e colunas para 4x2
```

## Ao remodelar o tensor, precisamos ter em mente que o número de elementos em seu interior deve permanecer o mesmo; por exemplo, se tínhamos 8 elementos no total, as possíveis formas para as quais ele pode ser convertido também devem conter 8 elementos.

```python
x.view(8,1) # 8 linhas e 1 coluna
```

## 2) Espremer
No squeeze, fornecemos o índice do eixo que queremos remover, o qual possui apenas um item nessa dimensão.

```python
a = torch.ones(2,1,10) # é criado um tensor em 2x1x10 e comprimimos a primeira dimensão
```

```python
#Este é um tensor composto por duas pilhas, cada uma com uma linha e 10 colunas, totalizando 20 elementos
a
```

```python
print(f"Este é o formato antes de apertar {a.shape}")
squeezed = a.squeeze(1)
print(f"Este é o formato depois de apertar {squeezed.shape}")
```

## Como você pode ver, removemos o primeiro índice do eixo, que contém apenas um item. Se o eixo tiver mais de um item, ele não será removido.

```python
a = torch.ones(2,2,10)
print(f"Este é o formato antes de apertar {a.shape}")
squeezed = a.squeeze(1)
print(f"Este é o formato depois de apertar {squeezed.shape}")
```

## 3) soltar
É o exato oposto de squeeze: em vez de remover, estamos adicionando uma dimensão ao nosso tensor.

```python
a = torch.ones(2,3)
print(f"Este é o formato antes de apertar {a.shape}")
squeezed = a.unsqueeze(2)
print(f"Este é o formato depois de apertar {squeezed.shape}")
```

## Indexação [None]: uma abordagem muito comum

```python
a = torch.ones(2,3)
```

```python
#Adicionar "none" criaria automaticamente uma dimensão fictícia no eixo especificado:
print(a[None].shape)# eixo fictício no índice 0
print(a[:,None].shape)#eixo fictício no índice 1
print(a[:,:,None].shape)#eixo fictício no índice 2
```

## Em visão computacional, tenha em mente que o formato do tensor seria:
Para tensor 1D(número total de elementos)

Para tensor 2D(linhas e colunas)

Para tensor 3D(número de canais, linhas e colunas)

Para tensor 4D(tamanho do lote, número de canais, linhas e colunas)
