```python
#bibliotecas
!pip install torch_snippets
from torch_snippets import *
from torchvision import transforms as T
from torch.nn import functional as F
from torchvision.models import vgg19

device = 'cuda' if torch.cuda.is_available() else 'cpu'
```

```python
# função de pré processamento, para normalizar a imagem
preprocess = T.Compose([
    T.ToTensor(), # converte a imagem em um tensor
    T.Normalize(mean=[0.485,0.456,0.406], std=[0.229,0.224,0.225]),
    T.Lambda(lambda x: x.mul_(255))])

#função de pós processamento, para desnormalizar a imagem
postprocess = T.Compose([
    T.Lambda(lambda x: x.mul_(1./255)), # escala a imagem
    T.Normalize(mean=[-0.485/0.229, -0.456/0.224,-0.406/0.225], std=[1/0.229,1/0.224,1/0.255]), ])
```

```python
# definição das três classes
class GramMatrix(nn.Module): # Ela calcula a matriz de Gram do tensor de entrada fornecido
    def forward(self,input):
        b,c,h,w = input.size()
        feat = input.view(b,c,h*w)
        G = feat@feat.transpose(1,2)
        G.div_(h*w)
        return G

class GramMSELoss(nn.Module): # Calcula o erro quadratico medio entre os traços gramaticais dos tensores de entrada
    def forward(self,input,target):
        out = F.mse_loss(GramMatrix()(input),target)
        return(out)

class vgg19_modified(nn.Module): # permiti extrair mapas de caracteristicas em camadas especificas
    def __init__(self):
        super().__init__()
        features = list(vgg19(pretrained=True).features)
        self.features = nn.ModuleList(features).eval()
    def forward(self, x, layers=[]):
        order = np.argsort(layers)
        _results, results = [], []
        for ix, model in enumerate(self.features):
            x = model(x)
            if ix in layers:
                _results.append(x)
        for o in order:
            results.append(_results[o])
        return results if layers is not [] else x
```

```python
vgg = vgg19_modified().to(device)
```

```python
# leitura das imagens
imgs = [Image.open(path).resize((512,512)).convert('RGB') for path in ['Fall_2022_web-images_Picasso_32.jpg','how-to-draw-an-elephant-featured-image-1200.png']]
style_image, content_image = [preprocess(img).to(device)[None] for img in imgs]
```

```python
#variavel de otimização
opt_img = content_image.data.clone()
opt_img.requires_grad = True
```

```python
style_layers = [0, 5, 10, 19, 28] # é definido as camadas de estilo
content_layers = [21] # é definido a camada de conteúdo(21)
loss_layers = style_layers + content_layers # camada sem perda como a concatenação das camadas de estilo e conteudo
```

```python
# é definido as funcoes de perda correspondente a cada camada de perda
loss_fns = [GramMSELoss()] * len(style_layers) + [nn.MSELoss()] * len(content_layers)
loss_fns = [loss_fn.to(device) for loss_fn in loss_fns]
```

```python
# É definido os pesos de cada camada de perda
style_weights = [1000/n**2 for n in [64,128,256,512,512]]
content_weights = [1]
weights = style_weights + content_weights
```

```python
style_target = [GramMatrix()(A).detach() for A in vgg(style_image, style_layers)]
content_targets = [A.detach() for A in vgg(content_image, content_layers)]
targets = style_target + content_targets
```

```python
# é definido um objeto de log para registrar a perda e exibir o processo de otimização
max_iters = 500
optimizer = optim.LBFGS([opt_img])
log = Report(max_iters)
```

```python
# é definido uma função de fechamento que calcula a perda e seus gradientes e atualiza a variavel de otimização
iters = 0
while iters < max_iters:
    def closure(): # função de passo para o fechamento
        global iters
        iters += 1
        optimizer.zero_grad()
        out = vgg(opt_img, loss_layers)
        layer_losses = [weights[a] * loss_fns[a](A, targets[a]) for a,A in enumerate(out)]
        loss = sum(layer_losses)
        loss.backward()
        log.record(pos=iters, loss=loss, end='\r')
        return loss # retorna
    optimizer.step(closure) # aplica o passo de otimização
```
