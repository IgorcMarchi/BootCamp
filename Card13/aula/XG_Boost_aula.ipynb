```python
#É importado o dataset da flor iris
from sklearn.datasets import load_iris

iris = load_iris() #é carregado o data_set

numSamples, numFeatures = iris.data.shape #pega as colunas do data_set
print(numSamples)
print(numFeatures)
print(list(iris.target_names)) #printa o nome e o numero de cada um
```

```python
#import
from sklearn.model_selection import train_test_split

# é separado em 20% dos dados para teste e os 80% para treino
X_train, X_test, y_train, y_test = train_test_split(iris.data, iris.target, test_size = 0.2, random_state = 0)
```

```python
#import do xgboost
import xgboost as xgb

train = xgb.DMatrix(X_train, label = y_train) #Os dados do treino são jogados para matriz D do import de cima
test = xgb.DMatrix(X_test, label = y_test) #mesma coisa se repete agora, porém para os dados de teste
```

```python
param = {
    'max_depth' : 4,
    'eta' : 0.3,
    'objective' : 'multi:softmax',
    'num_class' : 3} #dicionario com as caracteristicas pra treinar o modelo
epochs = 10 #quantidade de vezes que ele vai repetir o treinamento para aprender
```

```python
model = xgb.train(param, train, epochs) #treina o modelo com os parametros determinado acima
```

```python
predictions = model.predict(test) #Aqui é testado ele com os dados de test
```

```python
print(predictions) #printa os resultados
```

```python
#import de acuracia
from sklearn.metrics import accuracy_score

accuracy_score(y_test, predictions) #faz a acuracia do teste
```
