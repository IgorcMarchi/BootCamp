```python
import numpy as np
import matplotlib.pyplot as plt #imports
```

```python
pageSpeeds = np.random.normal(3.0, 1.0, 1000) # Velocidades de página aleatórias
purchaseAmount = 100 -  (pageSpeeds + np.random.normal(0, 0.1, 1000)) * 3 # Valores de compra simulados

plt.scatter(pageSpeeds, purchaseAmount) # gráfico de dispersão
plt.show()
```

```python
from scipy import stats
# Calcula regressão linear
slope, intercept, r_value, p_value, std_err = stats.linregress(pageSpeeds, purchaseAmount)
```

```python
r_value ** 2 #valor R ao quadrado
```

```python
def predict(x):  # Calcula valor previsto
    return slope * x + intercept # Equação da reta

fitLine = predict(pageSpeeds) # Gera valores da reta

plt.scatter(pageSpeeds, purchaseAmount) # Plota os pontos
plt.plot(pageSpeeds, fitLine, c ='r') #faz a linha vermelha
plt.show()
```
