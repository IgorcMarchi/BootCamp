```python
#imports
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
from sklearn.preprocessing import scale
import numpy as np
import random

def createClusterData(N, k): #funcao de cluster data
    np.random.seed(10) #seed pra não ficar diferente do video
    pointsPerCluster = float(N) / k  #divide as a quantidade de grupo por N
    X = [] #lista coemça vazia
    for i in range(k): #for que vai para cada grupo
        incomeCentroid = np.random.uniform(20000.0, 200000.0) #Randomiza o centroide entre 20 mil e 200 mil pra fazer o grupo
        ageCentroid = np.random.uniform(20.0, 70.0) #randomiza a idade do grupo
        for j in range(int(pointsPerCluster)): #pra cada grupo ele repete a quantidade que deu na divisao
            X.append([np.random.normal(incomeCentroid, 10000.0), np.random.normal(ageCentroid, 2.0)]) #gera renda com desvio padrão de 10000 o centro e a media do grupo mesma coisa com a idade
    X = np.array(X) #Envia para um array o conjunto de dados de idade e renda agrupados
    return X # retorna o valor
```

```python
data = createClusterData(100, 5) #utiliza a função de criar cluester de 100 pessoas em 5 grupos

model = KMeans(n_clusters = 5) #modelo com 5 grupos

model = model.fit(scale(data)) #treina o modelo

print(model.labels_) #mostra os numeros embaixo
plt.figure(figsize = (8, 6)) #mostra a figurafigura
plt.scatter(data[:, 0], data[:, 1], c = model.labels_.astype(float)) #mostra dados e cores
plt.show() # exibe
```
